```python

import pandas as pd
df = pd.read_csv("hf://datasets/lukebarousse/data_jobs/data_jobs.csv")


df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

## 🟩 Bar Chart with Labels (1.30.1)

```python
df.info()
```

```python
df['job_posted_month']=df['job_posted_date'].dt.month
graph = df.groupby('job_title_short')['salary_year_avg'].median().sort_values(ascending=False)
graph.plot(kind='bar', xlabel='Job Title', ylabel='US dollars', title='Job Title by Yearly Salary')
```

---

## 🟨 Horizontal Bar Chart with Labels (1.30.2)

```python
import matplotlib.pyplot as plt
plt.barh(graph.index, graph)
plt.xlabel('US dollars')
plt.ylabel('Job Title')
plt.title('Job Title by Yearly Salary')
plt.show()
```

---

## 🟥 Line Chart with Labels (1.30.3)

```python
plt.plot(graph.index, graph)
plt.xlabel('Job Title')
plt.xticks(rotation=90)
plt.ylabel('US dollars')
plt.title('Job Title by Yearly Salary')
plt.show()
```
