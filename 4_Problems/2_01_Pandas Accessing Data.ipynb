```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

## 🟩 Select First 5 Rows with .iloc() (2.1.1)

```python
df.iloc[0:5]
```

---

## 🟩 Get Rows and Columns with .loc() (2.1.2)

```python
df.loc[: , :]
```

---

## 🟨 First 15 Rows in US (2.1.3)

```python
df.loc[(df['job_country']=='United States')].head(15)
```

## 🟥 Filter by Date (2.1.4)

```python
df.sort_values(by='job_posted_date', ascending=True)
```
