```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
df = pd.read_csv("hf://datasets/lukebarousse/data_jobs/data_jobs.csv")

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

## 🟩 Remove Rows with Missing Values (2.2.1)

```python
df
```

```python
df.info()
```

```python
df = df.dropna()
```

```python
df.head()
```

---

## 🟨 Remove Duplicate Rows (2.2.2)

```python
df = df.drop_duplicates()
```

---

## 🟨 Fill Missing Values in Column (2.2.3)

```python
median_yearly = df.salary_year_avg.median()
median_hourly = df.salary_hour_avg.median()

df['salary_year_avg'] = df.salary_year_avg.fillna(median_yearly)
df['salary_hour_avg'] = df.salary_hour_avg.fillna(median_hourly)
df = df.median_hourly.drop()
```

```python
df.info()
```

```python
df = df.drop(columns='median_hourly')
```

```python
df.plot(x='job_title_short', y='salary_year_avg', kind='bar')
```

```python
df.groupby('job_title_short')['salary_year_avg'].agg(['median', 'mean', 'max', 'min'])
```
